```python
import pandas as pd
import os
```

```python
data = []

for f in os.listdir("C://Data//PseudoPFLOW//activity_v2//12"):
    df = pd.read_csv("C://Data//PseudoPFLOW//activity_v2//12//"+f, names=['pid', 'age','gender','occupation','starttime','duration', 'purpose', 'lon', 'lat', 'address'])
    data.append(df)

chiba_act = pd.concat(data, axis=0)
chiba_act

【139.92300735680504， 139.96621967375887】 lat【35.88552717539729, 35.9117804553585】,的部分
```

```python
kashiwa_act = chiba_act[(chiba_act['lon']<139.96621967375887)&(chiba_act['lon']>139.92300735680504)&(chiba_act['lat']<35.9117804553585)&(chiba_act['lat']>35.88552717539729)]
len(kashiwa_act['pid'].unique())
```

```python
kashiwa_act_final = chiba_act[chiba_act['pid'].isin(kashiwa_act['pid'].unique().tolist())]
kashiwa_act_final
```

```python
kashiwa_act_final.to_csv('kashiwa_act_final.csv', index=False, header=False)
```
